```python
!pip install groq --quiet
from groq import Groq
from google.colab import userdata
import os
import pandas as pd
```

```python
os.environ["GROQ_API_KEY"] = userdata.get("Groq_API_key")
client = Groq()
```

```python
df = pd.read_csv('reviews.csv')
```

```python
reviews = df['reviewText']
feedback = []
```

```python
feedback = []

for review in reviews:
    completion = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {
                "role": "user",
                "content": f"Classifique essa review como positiva, negativa ou neutra. Só digite uma das 3 e mais nada: {review}"
            }
        ],
        temperature=0,
        max_completion_tokens=10
    )

    resposta = completion.choices[0].message.content
    feedback.append(resposta)
```

```python
df['Feedback'] = feedback
```

```python
df_reviews_negativas = df[df['Feedback'] == 'Negativa']
```

```python
reviews_negativas = df_reviews_negativas['reviewText']
```

```python
texto_reviews_negativas = "###".join(reviews_negativas)
```
