```python
from google import genai

client = genai.Client()

while True:
  pergunta = input("Digite sua pergunta ou x para encerrar conversa: ")
  if pergunta == "x":
    break
  response = client.models.generate_content(
      model="gemini-3-flash-preview", contents=pergunta
  )
  print(response.text)
```

Salvando o histórico de conversas com o chatbot:

```python
chat = client.chats.create(model="gemini-3-flash-preview")

while True:
  pergunta = input("Digite sua pergunta ou x para encerrar: ")
  if pergunta == "x":
    break
  resposta = chat.send_message(pergunta)
  print(resposta.text)
```

Estruturas de dados:

```python
medias_alunos = {
    "Maria Silva": 8.9,
    "João Santos": 7.5,
    "Ana Oliveira": 4.2,
    "Pedro Costa": 1.4,
    "Camila Rodrigues": 9.5
}

print(medias_alunos.get("Pedro Costa"))
print(medias_alunos.keys())
print(medias_alunos.values())
print(medias_alunos.items())
```

```python
for nome, media in medias_alunos.items():
  print(f"Situação de {nome}:")

  media += 1
  if media < 5:
    print(f"Sua nota final foi {media}. Reprovado(a).")
  elif media < 10:
    print(f"Sua nota final foi {media}. Aprovado(a)!")
  else:
    print("Sua nota foi 10! Aprovado(a)!")

  print()
```

Forma menos pythonica:

```python
alunos = [
    "Maria Silva",
    "João Santos",
    "Ana Oliveira",
    "Pedro Costa",
    "Camila Rodrigues"
]

medias = [8.9, 7.5, 4.2, 1.4, 9.5]
```

```python
contador = 0

while contador <= 4:
  print(f"Situação de {alunos[contador]}:")

  medias[contador] += 1
  if medias[contador] < 5:
    print(f"Sua nota final foi {medias[contador]}. Reprovado(a).")
  elif medias[contador] < 10:
    print(f"Sua nota final foi {medias[contador]}. Aprovado(a)!")
  else:
    print("Sua nota foi 10! Aprovado(a)!")

  print()
  contador += 1
```
