## Lendo e Escrevendo Arquivos

```python
!pip install groq --quiet
import os
from groq import Groq
from google.colab import userdata
```

```python
os.environ["GROQ_API_KEY"] = userdata.get("Groq_API_key")
client = Groq()
```

```python
emails_escola = [
  "Prezado(a) responsável, informamos que houve uma reunião extraordinária do conselho de classe hoje. Mais detalhes serão enviados por e-mail nos próximos dias.",
  "Aviso importante: As aulas da próxima terça-feira, dia 15/05, serão suspensas devido a uma manutenção emergencial na escola. Pedimos desculpas pelo transtorno.",
  "Lembrete: O prazo final para a entrega do projeto de ciências é nesta sexta-feira. Certifiquem-se de que todos os materiais estão prontos. Em caso de dúvida, entre em contato com o professor."
]
```

```python
def resumir_emails(mensagens_escola: list[str]) -> list[str]:
  lista_resumos = []

  for numero, email_content in enumerate(emails_escola, start=1):
      completion = client.chat.completions.create(
          model="llama-3.1-8b-instant",
          messages=[
              {
                  "role": "user",
                  "content": f"Resuma este e-mail em poucas palavras: {email_content}"
              }
          ],
          temperature=0,
      )

      resumo = completion.choices[0].message.content
      lista_resumos.append(f"E-mail {numero}: {resumo}")

  return lista_resumos
```

```python
lista_resumos = resumir_emails(emails_escola)
```

```python
with open("lista_resumo.txt", "w", encoding="utf-8") as arquivo:
  for resumo in lista_resumos:
    arquivo.write(resumo + "\n")
```

```python
with open("lista_resumo.txt", "r",  encoding="utf-8") as arquivo:
  texto = arquivo.read()
  print(texto)
```

Desafio com pandas:

```python
perguntas = [
    "Qual é o planeta mais próximo da Terra?",
    "Quem é o autor mais lido no mundo?",
    "Qual a distância entre a Lua e a Terra?",
    "De onde vem o plástico?",
    "Qual é o instrumento mais bonito?"
    ]
```

```python
with open("perguntas.txt", "w", encoding="utf-8") as arquivo:
  for pergunta in perguntas:
    arquivo.write(pergunta + "\n")
```

```python
perguntas = []

with open("perguntas.txt", "r", encoding="utf-8") as arquivo:
  for linha in arquivo:
    perguntas.append(linha.strip())
```

```python
respostas = []

for pergunta in perguntas:
    completion = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {
                "role": "user",
                "content": f"Responda isso com apenas uma palavra: {pergunta}"
            }
        ],
        temperature=0,
    )

    resposta = completion.choices[0].message.content
    respostas.append(resposta)
```

```python
print(respostas)
```

```python
perguntas_e_respostas = dict(zip(perguntas, respostas))
print(perguntas_e_respostas)
```

```python
with open("resultados.csv", "w", encoding="utf-8") as arquivo:
  arquivo.write("Perguntas, Respostas\n")
  for pergunta, resposta in perguntas_e_respostas.items():
    arquivo.write(f"{pergunta}, {resposta}\n")
```
