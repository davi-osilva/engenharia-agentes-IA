```python
pessoa = {
"nome": "Gustavo",
"idade": 19,
"altura": 1.89
}
```

Utilizando o range:

```python
# o range começa a contagem em 0 e exclui o último número
for i in range(7):
  print(i, end=" ")
```

```python
# para imprimir de 1 à 7:
for i in range(1, 8):
  print(i, end=" ")
```

```python
# início, fim e etapas - a sequência que o Python utiliza
for i in range(2, 21, 2):
  print(i, end=" ")
```

Aprendendo funções:

```python
nomes_baguncados = [
    '  Joao    AlmeIDA ',
    'maRIA  SAntANA',
    '  PeDro OliveiRA',
    'AnA   JulIA ',
    'lUIz PerEIRA   ']
```

```python
def corrigir_nomes(nomes_baguncados: list[str]) -> list[str]:
  nomes_corretos = []

  for nome in nomes_baguncados:
    nome = nome.split()
    nome = " ".join(nome).title()
    nomes_corretos.append(nome)

  return nomes_corretos
```

```python
def main():
  nomes_corretos = corrigir_nomes(nomes_baguncados)

  for nome in nomes_corretos:
    print(nome)
```

```python
main()
```

Trabalhando com aleatoriedade:

```python
import random
```

```python
salas = ["sala X", "Sala Y", "Sala Z"]
alunos = [
  "Joao Almeida",
  "Maria Santana",
  "Pedro Oliveira",
  "Ana Julia",
  "Luiz Pereira"]
```

```python
sala_alunos = {}
```

```python
for aluno in alunos:
  sala = random.choice(salas)
  sala_alunos[aluno] = sala
```

```python
for aluno, sala in sala_alunos.items():
  print(f"{aluno}: {sala}")
```

```python
emails_escola = [
  "Prezado(a) responsável, informamos que houve uma reunião extraordinária do conselho de classe hoje. Mais detalhes serão enviados por e-mail nos próximos dias.",
  "Aviso importante: As aulas da próxima terça-feira, dia 15/05, serão suspensas devido a uma manutenção emergencial na escola. Pedimos desculpas pelo transtorno.",
  "Lembrete: O prazo final para a entrega do projeto de ciências é nesta sexta-feira. Certifiquem-se de que todos os materiais estão prontos. Em caso de dúvida, entre em contato com o professor."
]
```

```python
def resumir_emails(mensagens_escola: list[str]) -> list[str]:
  for numero, email_content in enumerate(mensagens_escola, start=1):
    resumo = client.models.generate_content(
      model="gemini-3-flash-preview",
      contents=f"Resuma este e-mail em poucas palavras: {email_content}")
    print(f"E-mail {numero}: {resumo.text}")
    print()
    print("-" * 100)
```

```python
os.environ["GROQ_API_KEY"] = userdata.get("Groq_API_key")
```

```python
from groq import Groq

client = Groq()
completion = client.chat.completions.create(
    model="meta-llama/llama-4-scout-17b-16e-instruct",
    messages=[
      {
        "role": "user",
        "content": "Quem ganha a Copa do Mundo de 2026?"
      }
    ],
    temperature=1,
    max_completion_tokens=1024,
    top_p=1,
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```
