Formatando textos:

```python
texto = "  Fabrício CARRaro  da Alura  "
print(texto)
```

```python
print(texto.lower())
print(texto.upper())
print(texto.strip())
print(texto.replace("  ", " "))
```

```python
print(texto)
```

Utilizando estruturas condicionais:

```python
idade = int(input("Digite sua idade: "))
```

```python
if idade < 18:
  print(f"Você não pode entrar na festa pois tem {idade} anos, logo, é menor de idade")
else:
  print(f"Você é maior de idade e pode entrar na festa, pois tem {idade} anos")
```

```python
media = float(input("Digite a sua média: "))
```

```python
if media < 5:
  print("Reprovado")
elif 5 <= media < 7:
  print("Recuperação")
else:
  print("Aprovado!")
```
